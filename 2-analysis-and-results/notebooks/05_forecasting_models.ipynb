# 05 - Forecasting Models
Train simple forecasting models (GradientBoosting) and save metrics.

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import os
os.makedirs('analysis/results', exist_ok=True)

# Load monthly aggregated Olist (if not present, user should run processing)
try:
    olist = pd.read_csv('1-Datasets/Processed_Data/monthly_sales.csv', parse_dates=['year_month'])
    olist = olist.sort_values('year_month')
    olist['month_num'] = range(1, len(olist) + 1)
except Exception as e:
    print('monthly_sales.csv not available or malformed:', e)
    raise

X = olist[['month_num']].values
y = olist['revenue'].values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
model = GradientBoostingRegressor(n_estimators=50, learning_rate=0.1, max_depth=3)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
results = pd.DataFrame({'Model':['Revenue'],'R2_Score':[r2],'MAE':[mae],'RMSE':[rmse]})
results.to_csv('analysis/results/forecast_results.csv', index=False)
print('Saved: analysis/results/forecast_results.csv')
```
