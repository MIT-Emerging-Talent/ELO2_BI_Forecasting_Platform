# 02 - Olist Processing & Validation
Clean and validate the merged Olist dataset, create features, and save cleaned outputs.

```python
import pandas as pd
import os
os.makedirs('analysis/results', exist_ok=True)

master = pd.read_csv('1-Datasets/Processed_Data/master_dataset.csv', parse_dates=['order_purchase_timestamp'])
print('Loaded master:', master.shape)

# Basic quality checks
print('Duplicates order_id:', master['order_id'].duplicated().sum())
print('Total missing values:', master.isnull().sum().sum())

# Filter extreme prices and create features
master = master[(master['price'] > 0) & (master['price'] < 10000)].copy()
master['year'] = master['order_purchase_timestamp'].dt.year
master['month'] = master['order_purchase_timestamp'].dt.month
master['year_month'] = master['order_purchase_timestamp'].dt.to_period('M')

master.to_csv('analysis/results/master_dataset_cleaned.csv', index=False)
print('Saved: analysis/results/master_dataset_cleaned.csv')
```
