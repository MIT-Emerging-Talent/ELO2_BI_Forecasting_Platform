# 01 - Exploratory Data Analysis
This notebook loads the processed datasets and produces basic statistics and visualizations. Run cells in order.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load processed data
olist = pd.read_csv('1-Datasets/Processed_Data/master_dataset.csv')
superstore = pd.read_csv('1-Datasets/Processed_Data/superstore_clean.csv')
# economic file may be missing until GDP is added
try:
    economic = pd.read_csv('1-Datasets/Processed_Data/economic_indicators_combined.csv')
except FileNotFoundError:
    economic = None

print('Olist rows,cols:', olist.shape)
print('Superstore rows,cols:', superstore.shape)
print('Economic loaded:', economic is not None)
```

```python
# Quick missing value summary
print('
Missing values summary:')
print('Olist missing:', olist.isnull().sum().sum())
print('Superstore missing:', superstore.isnull().sum().sum())
if economic is not None:
    print('Economic missing:', economic.isnull().sum().sum())
```

```python
# Simple visualizations (saved to analysis/results/)
import os
os.makedirs('analysis/results', exist_ok=True)
plt.figure(figsize=(8,4))
plt.hist(olist['price'].dropna(), bins=50, color='teal')
plt.title('Olist price distribution')
plt.xlabel('Price')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('analysis/results/01_exploratory_price_dist.png', dpi=150)
print('Saved: analysis/results/01_exploratory_price_dist.png')
```
