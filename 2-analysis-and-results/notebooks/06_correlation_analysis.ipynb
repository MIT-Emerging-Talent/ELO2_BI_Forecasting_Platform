# 06 - Correlation Analysis
Compute Pearson correlations between revenue/orders and economic indicators.

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
import os
os.makedirs('analysis/results', exist_ok=True)

olist = pd.read_csv('1-Datasets/Processed_Data/monthly_sales.csv', parse_dates=['year_month'])
econ = pd.read_csv('1-Datasets/Processed_Data/brazil_inflation.csv')
# Simplify econ to monthly annual_inflation_rate by date
econ_month = econ[['date','annual_inflation_rate']].copy()
econ_month['date'] = pd.to_datetime(econ_month['date'])
econ_month['year_month'] = econ_month['date'].dt.to_period('M').dt.to_timestamp()
merged = olist.merge(econ_month[['year_month','annual_inflation_rate']], left_on='year_month', right_on='year_month', how='left')
merged = merged.dropna(subset=['annual_inflation_rate'])
corr, p = pearsonr(merged['revenue'], merged['annual_inflation_rate'])
out = pd.DataFrame({'Variable Pair':['Revenue vs Inflation'],'Correlation':[corr],'P_Value':[p]})
out.to_csv('analysis/results/correlation_analysis.csv', index=False)
print('Saved: analysis/results/correlation_analysis.csv')
```
