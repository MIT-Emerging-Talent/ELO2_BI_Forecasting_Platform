# 04 - Economic Indicators
Analyze inflation & GDP and create visualizations and summaries.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
os.makedirs('analysis/results', exist_ok=True)

try:
    econ = pd.read_csv('1-Datasets/Processed_Data/economic_indicators_combined.csv')
except FileNotFoundError:
    print('economic_indicators_combined.csv not found; create it by running data processing script')
    econ = None

if econ is not None:
    print(econ.head())
    plt.figure(figsize=(8,3))
    plt.plot(econ['date'], econ['annual_inflation_rate'], marker='o')
    plt.title('Annual inflation rate')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('analysis/results/04_economic_inflation.png', dpi=150)
    print('Saved: analysis/results/04_economic_inflation.png')
```
