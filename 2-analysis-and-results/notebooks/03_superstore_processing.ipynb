# 03 - Superstore Processing
Clean, analyze, and save Superstore summaries.

```python
import pandas as pd
import os
os.makedirs('analysis/results', exist_ok=True)

ss = pd.read_csv('1-Datasets/Processed_Data/superstore_clean.csv')
print('Loaded superstore:', ss.shape)

ss['Order Date'] = pd.to_datetime(ss['Order Date'], errors='coerce')
ss['year_month'] = ss['Order Date'].dt.to_period('M')

category = ss.groupby('Category').agg(Sales=('Sales','sum'), Profit=('Profit','sum')).reset_index()
category.to_csv('analysis/results/superstore_category_profitability.csv', index=False)
print('Saved: analysis/results/superstore_category_profitability.csv')
```
